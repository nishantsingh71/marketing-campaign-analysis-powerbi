## Feature Engineering

In this section, new features are created from the existing data to enhance business analysis and improve the overall understanding of marketing campaign performance. These engineered features provide meaningful business metrics that help evaluate campaign effectiveness, profitability, and customer engagement.

The following features are created:

- **`Profit`** – Calculate the profit earned from each campaign.
- **`Profit Margin (%)`** – Calculate the percentage of profit generated from revenue.
- **`ROI (%)`** – Measure the return on investment for each campaign.
- **`Campaign Duration (Days)`** – Calculate the total duration of each campaign.
- **`CTR (%)`** – Calculate the Click-Through Rate to measure ad engagement.
- **`Conversion Rate (%)`** – Calculate the percentage of clicks converted into successful conversions.

```python
# Create a new 'Profit' column
# Profit is calculated by subtracting Spend from Revenue.
df["Profit"] = df["Revenue"] - df["Spend"]
```

```python
# Create a new 'Profit Margin (%)' column
# It shows the percentage of profit earned from revenue.
df["Profit Margin (%)"] = (df["Profit"]/df["Revenue"])*100
```

```python
# Create a new 'ROI (%)' column
# ROI measures the return generated for every unit of campaign spend.
df["ROI (%)"] = (df["Profit"]/df["Spend"])*100
```

```python
# Create a new 'Campaign Duration (Days)' column
# Calculate the total duration of each campaign in days.
df["Campaign Duration (Days)"]= (df["End_Date"] - df["Start_Date"]).dt.days
```

```python
# Create a new 'CTR (%)' column
# CTR (Click-Through Rate) measures the percentage of users who clicked the ad after seeing it.
df["CTR (%)"] = (df["Clicks"]/df["Impressions"])*100
```

```python
# Create a new 'Conversion Rate (%)' column
# It measures the percentage of clicks that resulted in conversions.
df["Conversion Rate (%)"] = (df["Conversions"] / df["Clicks"]) * 100
```

```python
df.head()
```

```python
# Display campaigns with negative duration
# This helps identify records where End_Date is earlier than Start_Date.
df[df["Campaign Duration (Days)"] < 0]
```

```python
# Count records where End_Date is earlier than Start_Date
# These records require date correction.
(df["End_Date"] < df["Start_Date"]).sum()
```

### Date Sequence Correction

Some records contained an invalid date sequence where the **End Date** occurred before the **Start Date**. These records were corrected by swapping the date values to maintain a valid campaign timeline before calculating the campaign duration.

```python
# Correct records where End_Date is earlier than Start_Date
# Swap the Start_Date and End_Date values to maintain a valid campaign timeline.
mask = df["End_Date"] < df["Start_Date"]

df.loc[mask, ["Start_Date", "End_Date"]] = (
    df.loc[mask, ["End_Date", "Start_Date"]].values
)
```

```python
(df["End_Date"] < df["Start_Date"]).sum()
```

```python
# Recalculate campaign duration after correcting the date sequence
# Calculate the total duration of each campaign in days.

df["Campaign Duration (Days)"] = (
    df["End_Date"] - df["Start_Date"]
).dt.days
```

```python
# Verify that no negative campaign durations remain
# The output should be 0.

(df["Campaign Duration (Days)"] < 0).sum()
```

## 3. Monthly Revenue Trend

### Business Question

How did the total revenue change over different months?

### Objective

Analyze the monthly revenue trend to identify seasonal patterns and understand revenue performance over time.

```python
# Extract month name from Start_Date
# Create a new Month column for time-based analysis.

df["Month"] = df["Start_Date"].dt.month_name()
```

```python
# Create Month Number column
# Extract month number from Start_Date for chronological sorting.

df["Month Number"] = df["Start_Date"].dt.month
```

```python
# Calculate total revenue for each month
# Group the data by Month and calculate total revenue.

monthly_revenue = (
    df.groupby(["Month Number", "Month"])["Revenue"]
      .sum()
      .reset_index()
      .sort_values("Month Number")
)

monthly_revenue
```

```python
# Plot monthly revenue trend
# This chart shows how revenue changes over different months.

plt.figure(figsize=(10,5))
plt.plot(
    monthly_revenue["Month"],
    monthly_revenue["Revenue"],
    marker="o",
    linewidth=2
)
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```

### Observation

- Revenue varies across different months.
- Some months generated significantly higher revenue than others.
- The line chart clearly highlights peaks and dips in monthly performance.
- No consistent increasing or decreasing trend is observed throughout the year.

### Business Insight

Monthly revenue fluctuates throughout the year, indicating seasonal variations in campaign performance. The company can identify high-performing months to increase marketing investment and analyze low-performing months to improve future campaign strategies.
